```python
import os
import glob
import librosa

# Path to your extracted actor folders
data_path = "data/"
actor_directories = glob.glob(os.path.join(data_path, "Actor_*"))

print(f"Found {len(actor_directories)} actor directories.")

# Pick the first audio file to test
sample_files = glob.glob(os.path.join(actor_directories[0], "*.wav"))
sample_file = sample_files[0]

print(f"Testing audio file: {sample_file}")

# Load audio with librosa
y, sr = librosa.load(sample_file, sr=22050)
print(f"Sample Rate: {sr}, Audio Shape: {y.shape}")

# Parse RAVDESS filename convention: e.g., 03-01-06-01-02-01-12.wav
# Part 3 represents the emotion code
filename = os.path.basename(sample_file)
parts = filename.split("-")
emotion_code = parts[2]

emotion_map = {
    "01": "neutral",
    "02": "calm",
    "03": "happy",
    "04": "sad",
    "05": "angry",
    "06": "fearful",
    "07": "disgust",
    "08": "surprised"
}

print(f"Emotion Code: {emotion_code} -> Label: {emotion_map.get(emotion_code)}")
```

```python
import os
import glob
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path

# Fix working directory if needed
if os.getcwd().endswith("experiments"):
    os.chdir("..")

def save_mel_spectrogram(audio_path, output_image_path, sr=22050, n_mels=128):
    """
    Loads an audio file, computes its Mel-spectrogram, 
    and saves it as a clean image without borders or axes.
    """
    try:
        # Load audio
        y, sr = librosa.load(audio_path, sr=sr)
        
        # Compute Mel-spectrogram
        mel_spectrogram = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=n_mels)
        mel_spectrogram_db = librosa.power_to_db(mel_spectrogram, ref=np.max)
        
        # Plot and save as image
        fig, ax = plt.subplots(figsize=(4, 4), dpi=100)
        ax.axis('off')
        librosa.display.specshow(mel_spectrogram_db, sr=sr, x_axis=None, y_axis=None, ax=ax)
        
        os.makedirs(os.path.dirname(output_image_path), exist_ok=True)
        plt.savefig(output_image_path, bbox_inches='tight', pad_inches=0)
        plt.close(fig)
        return True
    except Exception as e:
        print(f"Error processing {audio_path}: {e}")
        return False

# Test it on a single file
sample_file = "data/Actor_01/03-01-01-01-01-01-01.wav"
output_test = "artifacts/feature_extraction/test_spectrogram.png"
success = save_mel_spectrogram(sample_file, output_test)

if success:
    print(f"Successfully generated spectrogram image at: {output_test}")
```
